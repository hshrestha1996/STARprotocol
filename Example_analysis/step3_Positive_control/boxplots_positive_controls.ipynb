```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_excel("Metainfo_with_covariate_sex_age_pmi_combined_correcteddata.xlsx",sheet_name="Sheet3")
```

```python
df.columns
```

```python
df.groupby("Diagnosis").count()
```

```python
df['Diagnosis'] = df['Diagnosis'].replace("ND", "Ctl")
```

```python
category_order = ['Ctl', 'AD', 'VAD']
df['Diagnosis'] = pd.Categorical(df['Diagnosis'], categories=category_order, ordered=True)
```

```python
for cols in df.columns[3:]:
    interest = cols
    ctl_mean = df[df['Diagnosis'] == 'Ctl'][interest].mean()

    # Set font properties
    plt.rcParams.update({
        'text.color': 'black',
        'axes.labelcolor': 'black',
        'xtick.color': 'black',
        'ytick.color': 'black'
    })

    plt.figure(figsize=(2, 2))

    sns.boxplot(x='Diagnosis', y=interest, data=df, color="white",boxprops=dict(edgecolor='black'), whiskerprops=dict(color='black'),
                medianprops=dict(color='black'), showfliers=False)
    sns.stripplot(x='Diagnosis', y=interest, data=df, color='grey', alpha=0.6, jitter=True, size=3)
    plt.axhline(ctl_mean, color='red', linestyle='--', zorder=10)

    # Adding title and labels
    plt.title(interest)
    plt.xlabel('')
    plt.ylabel('Log2 abundance')

    # Rotate x-axis tick labels by 90 degrees
    plt.gca().tick_params(axis='x', labelrotation=90)

    # Save plot
    plt.savefig(f'Boxplot_{interest}_VADcohort.png', dpi=300, bbox_inches='tight')

    # Show plot
    plt.show()
```
