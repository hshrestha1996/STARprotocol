```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv("../step2_covariate_correction/covariate_sex_age_pmi_combined_corrected.csv", index_col=0)
df=df.transpose()

geneinfo= pd.read_excel("protein_gene_description_table.xlsx", sheet_name= "Sheet2")
geneinfo.columns=['ProteinAccession', 'ProteinDescription', 'gene']
```

```python
df.reset_index(inplace=True)
df.rename(columns={'index': 'ProteinAccession'}, inplace=True)
```

```python
df1 =geneinfo.merge(df, on="ProteinAccession", how="right")
```

```python
df1
```

```python
marker = pd.read_table("./Markers_BRETIGEA_specific_top500.txt")
```

```python
markermatch=marker.merge(df1, on="gene", how="inner")
markermatch=markermatch[markermatch.grand_mean>3]
markermatch.drop_duplicates(subset="gene", inplace=True)
mar_copy= markermatch.drop(columns=["grand_mean", "gene", "ProteinAccession", "ProteinDescription"])
mar= markermatch.drop(columns=["grand_mean", "gene", "ProteinAccession", "ProteinDescription"])
```

```python
mar_copy.groupby("Celltype").count().transpose().head()
```

```python
mar=mar.groupby("Celltype").mean().transpose()
mar.columns= ['Average_ast_markerabundance', 'Average_end_markerabundance', 
              'Average_mic_markerabundance', 'Average_neu_markerabundance', 'Average_oil_markerabundance']
mar=mar.reset_index()
```

```python
meta = pd.read_csv("meta.csv")
```

```python
meta.columns
```

```python
meta.columns=['index', 'Batch_info', 'ApoE', 'Diagnosis', 'Sex', 'Age', 'PMI',
       'Cerad NP', 'Braak score', 'tmtinfo', 'Batch']
meta=meta.merge(mar, on="index", how="left")
meta=meta.sort_values("Diagnosis")
```
